# Hospital Appointment Data Analysis

**Subject:** Python Programming / Data Analysis (First Semester)  
**Platform:** Google Colab  

## Aim
To create and analyse a hospital appointment dataset of 40 patients using basic Python, NumPy, and Pandas.

## Objectives
1. Create a sample dataset of 40 hospital appointments using NumPy and Pandas.
2. Inspect the dataset structure, columns, and missing values.
3. Count total scheduled, completed, cancelled, and missed appointments.
4. Find department-wise and doctor-wise appointment counts.
5. Calculate average patient age and average waiting time.
6. Compare morning and afternoon appointment demand.

```python
# Step 1: Import NumPy and Pandas libraries
import numpy as np
import pandas as pd
```

## 1. Creating the Dataset (40 Patients)

We generate sample appointment data for 40 patients using NumPy random functions and store it in a Pandas DataFrame.

```python
# Step 2: Create sample data for 40 patients
np.random.seed(10)
n = 40
departments = ['General Medicine', 'Cardiology', 'Orthopedics', 'Pediatrics', 'Dermatology']
doctors = ['Dr. Sharma', 'Dr. Verma', 'Dr. Singh', 'Dr. Khan', 'Dr. Patel']
statuses = ['Completed', 'Completed', 'Completed', 'Cancelled', 'Missed']
shifts = ['Morning', 'Afternoon']
data = {
    'Patient_ID': np.arange(101, 101 + n),
    'Age': np.random.randint(10, 80, n),
    'Department': np.random.choice(departments, n),
    'Doctor': np.random.choice(doctors, n),
    'Shift': np.random.choice(shifts, n),
    'Status': np.random.choice(statuses, n),
    'Wait_Minutes': np.random.randint(10, 45, n)
}
df = pd.DataFrame(data)
df.to_csv('patient_appointments_dataset.csv', index=False)
```

## 2. Viewing and Checking the Dataset

We display the first five rows, dataset shape, column names, and check for any missing values.

```python
# Step 3: Inspect the dataset
print('First 5 rows of the dataset:')
print(df.head())
print('Rows and Columns:', df.shape)
print('Column Names:', list(df.columns))
print('Missing Values:')
print(df.isnull().sum())
```

## 3. Appointment Status Analysis

We count the total appointments and see how many were completed, cancelled, or missed.

```python
# Step 4: Total and status-wise appointment counts
total_appointments = len(df)
print('Total Scheduled Appointments:', total_appointments)
status_counts = df['Status'].value_counts()
print('Appointment Status Count:')
print(status_counts)
missed_count = (df['Status'] == 'Missed').sum()
missed_rate = round((missed_count / total_appointments) * 100, 1)
print('Overall Missed Rate (%):', missed_rate)
```

## 4. Department-wise and Doctor-wise Appointments

We find which hospital department and which doctor handled the most appointments.

```python
# Step 5: Department-wise and doctor-wise counts
dept_counts = df['Department'].value_counts()
print('Appointments by Department:')
print(dept_counts)
doctor_counts = df['Doctor'].value_counts()
print('Appointments by Doctor:')
print(doctor_counts)
```

## 5. Shift Demand, Age, and Waiting Time Analysis

We compare Morning vs. Afternoon appointments and calculate average patient age and average waiting time for completed visits.

```python
# Step 6: Shift demand, patient age, and waiting time
shift_counts = df['Shift'].value_counts()
print('Appointments by Shift:')
print(shift_counts)
avg_age = round(df['Age'].mean(), 1)
print('Average Patient Age:', avg_age)
completed_df = df[df['Status'] == 'Completed']
avg_wait = round(completed_df['Wait_Minutes'].mean(), 1)
print('Average Waiting Time (Completed Visits):', avg_wait, 'mins')
```

## 6. Final Summary

We print a clean summary of our key findings.

```python
# Step 7: Print key findings
print('--- KEY SUMMARY ---')
print('Total Patients:', total_appointments)
print('Busiest Department:', dept_counts.idxmax())
print('Busiest Shift:', shift_counts.idxmax())
print('Missed Rate (%):', missed_rate)
print('Average Wait (mins):', avg_wait)
```

## Conclusion

In this project, we analysed a sample hospital appointment dataset of 40 patients using basic Python, NumPy, and Pandas. We inspected the dataset, counted completed, cancelled, and missed visits, compared department and doctor workloads, and calculated average patient age and waiting times.
